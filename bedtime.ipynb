```python
import pandas as pd
import plotly.express as px


# Mark records with severe sleep debt
df["is_severe"] = (
    df["sleep_debt_category"] == "Severe Sleep Debt"
)

# Calculate the severe-debt rate within each occupation
summary = (
    df.groupby("occupation_type", as_index=False)
      .agg(
          records=("sleep_debt_category", "size"),
          severe_records=("is_severe", "sum")
      )
)

summary["severe_pct"] = (
    100 * summary["severe_records"] / summary["records"]
)

# Rank highest to lowest
summary = summary.sort_values(
    "severe_pct",
    ascending=False
)

# Format fields used in the tooltip
summary["severe_pct_label"] = summary["severe_pct"].map(
    lambda value: f"{value:.1f}%"
)

summary["record_count_label"] = summary.apply(
    lambda row: (
        f"{int(row['severe_records']):,} of "
        f"{int(row['records']):,}"
    ),
    axis=1
)

# Create the visualization
fig = px.bar(
    summary,
    x="severe_pct",
    y="occupation_type",
    orientation="h",
    custom_data=[
        "severe_pct_label",
        "severe_records",
        "records"
    ],
    labels={
        "severe_pct": "Records with severe sleep debt (%)",
        "occupation_type": "Occupation type"
    },
    title="Severe sleep debt is most common among healthcare and shift workers"
)

# Put the highest rate at the top
fig.update_yaxes(
    categoryorder="array",
    categoryarray=summary["occupation_type"].tolist(),
    autorange="reversed",
    title=None
)

# Use a shared zero-based percentage scale
fig.update_xaxes(
    range=[0, summary["severe_pct"].max() * 1.15],
    ticksuffix="%",
    rangemode="tozero",
    title="Records with severe sleep debt"
)

# Customize hover information
fig.update_traces(
    hovertemplate=(
        "<b>%{y}</b><br>"
        "Severe-debt rate: %{customdata[0]}<br>"
        "Severe records: %{customdata[1]:,}<br>"
        "Total records: %{customdata[2]:,}"
        "<extra></extra>"
    )
)

fig.update_layout(
    template="plotly_white",
    showlegend=False,
    margin=dict(l=190, r=40, t=80, b=60)
)

fig.show()

# Optional: inspect the underlying figures
print(
    summary[
        ["occupation_type", "severe_records", "records", "severe_pct"]
    ].to_string(index=False)
)
```

```python
# Desired display order
chronotype_order = [
    "Morning Lark",
    "Intermediate",
    "Night Owl"
]
debt_order = [
    "Optimal Recovery",
    "Mild Deficit",
    "Moderate Debt",
    "Severe Sleep Debt"
]

# Count records in each chronotype × sleep-debt combination
summary = (
    df.groupby(
        ["chronotype", "sleep_debt_category"],
        observed=False
    )
    .size()
    .reset_index(name="user_count")
)

# Calculate the number of users within each chronotype
summary["chronotype_n"] = (
    summary.groupby("chronotype")["user_count"]
    .transform("sum")
)

# Convert counts to percentages within chronotype
summary["user_pct"] = (
    100 * summary["user_count"] / summary["chronotype_n"]
)

# Apply the intended category order
summary["chronotype"] = pd.Categorical(
    summary["chronotype"],
    categories=chronotype_order,
    ordered=True
)

summary["sleep_debt_category"] = pd.Categorical(
    summary["sleep_debt_category"],
    categories=debt_order,
    ordered=True
)

summary = summary.sort_values(
    ["chronotype", "sleep_debt_category"]
)

# Create percentage labels for the tooltip
summary["pct_label"] = summary["user_pct"].map(
    lambda value: f"{value:.1f}%"
)

# Create the 100% stacked bar chart
fig = px.bar(
    summary,
    x="chronotype",
    y="user_pct",
    color="sleep_debt_category",
    category_orders={
        "chronotype": chronotype_order,
        "sleep_debt_category": debt_order
    },
    custom_data=[
        "sleep_debt_category",
        "user_count",
        "chronotype_n",
        "pct_label"
    ],
    labels={
        "chronotype": "Chronotype",
        "user_pct": "Users within chronotype (%)",
        "sleep_debt_category": "Sleep-debt category"
    },
    title="Sleep debt is highest among Night Owls and lowest among Morning Larks"
)

# Format the percentage axis
fig.update_yaxes(
    range=[0, 100],
    ticksuffix="%",
    title="Users within chronotype"
)

# Customize hover information
fig.update_traces(
    hovertemplate=(
        "<b>%{x}</b><br>"
        "Sleep-debt category: %{customdata[0]}<br>"
        "Share: %{customdata[3]}<br>"
        "Users in category: %{customdata[1]:,}<br>"
        "Total users in chronotype: %{customdata[2]:,}"
        "<extra></extra>"
    )
)

fig.update_layout(
    template="plotly_white",
    barmode="stack",
    legend_title_text="Sleep-debt category",
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="left",
        x=0
    ),
    margin=dict(l=70, r=30, t=120, b=60)
)

fig.show()

# Optional: calculate the combined moderate-or-severe percentage
combined = (
    df.assign(
        moderate_or_severe=df["sleep_debt_category"].isin(
            ["Moderate Debt", "Severe Sleep Debt"]
        )
    )
    .groupby("chronotype", as_index=False)
    .agg(
        total_users=("chronotype", "size"),
        moderate_or_severe_users=("moderate_or_severe", "sum")
    )
)

combined["moderate_or_severe_pct"] = (
    100
    * combined["moderate_or_severe_users"]
    / combined["total_users"]
)

combined["chronotype"] = pd.Categorical(
    combined["chronotype"],
    categories=chronotype_order,
    ordered=True
)

combined = combined.sort_values("chronotype")

print(
    combined[
        [
            "chronotype",
            "total_users",
            "moderate_or_severe_users",
            "moderate_or_severe_pct"
        ]
    ].to_string(index=False)
)
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import chi2_contingency



# Remove the identifier: correlations involving user_id are meaningless
analysis_df = df.drop(columns=["user_id"], errors="ignore").copy()

# Treat coded binary fields as categorical
categorical_columns = [
    "gender",
    "occupation_type",
    "chronotype",
    "primary_bedtime_app",
    "blue_light_filter_active",
    "sleep_debt_category"
]

categorical_columns = [
    column for column in categorical_columns
    if column in analysis_df.columns
]

numeric_columns = [
    column for column in analysis_df.columns
    if column not in categorical_columns
    and pd.api.types.is_numeric_dtype(analysis_df[column])
]

print("Numeric parameters:")
print(numeric_columns)

print("\nCategorical parameters:")
print(categorical_columns)


# ------------------------------------------------------------
# 2. Functions for mixed-variable associations
# ------------------------------------------------------------

def cramers_v(x, y):
    """
    Bias-corrected Cramer's V for two categorical variables.
    Returns an association strength between 0 and 1.
    """
    valid = pd.DataFrame({"x": x, "y": y}).dropna()

    if valid.empty:
        return np.nan

    confusion_matrix = pd.crosstab(valid["x"], valid["y"])

    if (
        confusion_matrix.empty
        or confusion_matrix.shape[0] < 2
        or confusion_matrix.shape[1] < 2
    ):
        return 0.0

    chi2 = chi2_contingency(
        confusion_matrix,
        correction=False
    )[0]

    n = confusion_matrix.to_numpy().sum()
    rows, columns = confusion_matrix.shape

    phi2 = chi2 / n

    # Bias correction
    corrected_phi2 = max(
        0,
        phi2 - ((columns - 1) * (rows - 1)) / (n - 1)
    )

    corrected_rows = rows - ((rows - 1) ** 2) / (n - 1)
    corrected_columns = columns - ((columns - 1) ** 2) / (n - 1)

    denominator = min(
        corrected_columns - 1,
        corrected_rows - 1
    )

    if denominator <= 0:
        return 0.0

    return np.sqrt(corrected_phi2 / denominator)


def correlation_ratio(categories, values):
    """
    Correlation ratio, eta, between a categorical variable
    and a numeric variable.

    Returns an association strength between 0 and 1.
    """
    valid = pd.DataFrame({
        "category": categories,
        "value": pd.to_numeric(values, errors="coerce")
    }).dropna()

    if valid.empty or valid["category"].nunique() < 2:
        return 0.0

    overall_mean = valid["value"].mean()

    between_group_variation = 0.0

    for _, group in valid.groupby("category", observed=False):
        between_group_variation += (
            len(group)
            * (group["value"].mean() - overall_mean) ** 2
        )

    total_variation = (
        (valid["value"] - overall_mean) ** 2
    ).sum()

    if total_variation == 0:
        return 0.0

    return np.sqrt(
        between_group_variation / total_variation
    )


def mixed_association(data, column_1, column_2, categorical):
    """
    Select an association measure based on variable types.

    Numeric vs numeric:
        Absolute Spearman correlation

    Categorical vs categorical:
        Bias-corrected Cramer's V

    Numeric vs categorical:
        Correlation ratio, eta
    """
    if column_1 == column_2:
        return 1.0

    first_is_categorical = column_1 in categorical
    second_is_categorical = column_2 in categorical

    if not first_is_categorical and not second_is_categorical:
        result = data[column_1].corr(
            data[column_2],
            method="spearman"
        )

        return abs(result) if pd.notna(result) else np.nan

    if first_is_categorical and second_is_categorical:
        return cramers_v(
            data[column_1],
            data[column_2]
        )

    if first_is_categorical:
        return correlation_ratio(
            data[column_1],
            data[column_2]
        )

    return correlation_ratio(
        data[column_2],
        data[column_1]
    )


# ------------------------------------------------------------
# 3. Calculate the all-parameter association matrix
# ------------------------------------------------------------
all_columns = analysis_df.columns.tolist()

association_matrix = pd.DataFrame(
    index=all_columns,
    columns=all_columns,
    dtype=float
)

for column_1 in all_columns:
    for column_2 in all_columns:
        association_matrix.loc[column_1, column_2] = (
            mixed_association(
                analysis_df,
                column_1,
                column_2,
                categorical_columns
            )
        )


# ------------------------------------------------------------
# 4. Plot the all-parameter association heatmap
# ------------------------------------------------------------
plt.figure(figsize=(18, 14))

sns.heatmap(
    association_matrix,
    cmap="YlOrRd",
    vmin=0,
    vmax=1,
    annot=True,
    fmt=".2f",
    linewidths=0.4,
    square=True,
    cbar_kws={
        "label": "Association strength (0–1)"
    }
)

plt.title(
    "Association Between All Sleep and Bedtime Parameters",
    fontsize=16,
    pad=20
)

plt.xticks(
    rotation=45,
    ha="right",
    fontsize=9
)

plt.yticks(
    rotation=0,
    fontsize=9
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. Rank the strongest parameter pairs
# ------------------------------------------------------------
pairs = []

for i, column_1 in enumerate(all_columns):
    for column_2 in all_columns[i + 1:]:
        pairs.append({
            "parameter_1": column_1,
            "parameter_2": column_2,
            "association_strength":
                association_matrix.loc[column_1, column_2]
        })

strongest_pairs = (
    pd.DataFrame(pairs)
    .dropna(subset=["association_strength"])
    .sort_values(
        "association_strength",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nStrongest parameter associations:")
print(strongest_pairs.head(20).to_string(index=False))


# ------------------------------------------------------------
# 6. Signed correlations for numeric and ordinal variables
# ------------------------------------------------------------
signed_df = analysis_df[numeric_columns].copy()

# Convert sleep-debt categories to an ordered numeric scale
if "sleep_debt_category" in analysis_df.columns:
    sleep_debt_order = {
        "Optimal Recovery": 0,
        "Mild Deficit": 1,
        "Moderate Debt": 2,
        "Severe Sleep Debt": 3
    }

    signed_df["sleep_debt_severity"] = (
        analysis_df["sleep_debt_category"]
        .map(sleep_debt_order)
    )

# Blue-light filter is binary, so it can also be included
if "blue_light_filter_active" in analysis_df.columns:
    signed_df["blue_light_filter_active"] = pd.to_numeric(
        analysis_df["blue_light_filter_active"],
        errors="coerce"
    )

spearman_matrix = signed_df.corr(method="spearman")

# Hide the duplicated upper half of the matrix
mask = np.triu(
    np.ones_like(spearman_matrix, dtype=bool),
    k=1
)

plt.figure(figsize=(16, 13))

sns.heatmap(
    spearman_matrix,
    mask=mask,
    cmap="RdBu_r",
    center=0,
    vmin=-1,
    vmax=1,
    annot=True,
    fmt=".2f",
    linewidths=0.4,
    square=True,
    cbar_kws={
        "label": "Spearman correlation"
    }
)

plt.title(
    "Signed Correlations Between Numeric Sleep Parameters",
    fontsize=16,
    pad=20
)

plt.xticks(
    rotation=45,
    ha="right",
    fontsize=9
)

plt.yticks(
    rotation=0,
    fontsize=9
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. Correlations specifically with next-day fatigue
# ------------------------------------------------------------
if "next_day_fatigue_score" in spearman_matrix.columns:
    fatigue_correlations = (
        spearman_matrix["next_day_fatigue_score"]
        .drop("next_day_fatigue_score")
        .sort_values(key=abs, ascending=False)
        .rename("spearman_correlation")
        .reset_index()
        .rename(columns={"index": "parameter"})
    )

    print("\nParameters most strongly correlated with next-day fatigue:")
    print(fatigue_correlations.to_string(index=False))
```

### Fixing the Data Preparation and Modeling Pipeline
Now that `df` has been properly processed with `pd.get_dummies()`, we will recreate the features `x` (excluding target variables) and target `y` (original categorical column), perform the train/test split, scale the features, and fit the model.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# 1. Define y using the original categorical column from the original dataset (reloaded to keep y clean)
original_df = pd.read_csv("bsdc.csv")
y = original_df['sleep_debt_category']

# 2. Define x by dropping any target-related dummy columns and the original target columns from our encoded df
target_dummy_cols = [col for col in df.columns if 'sleep_debt_category' in col]
x = df.drop(columns=target_dummy_cols, errors='ignore')

# 3. Split the data
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# 4. Standardize the features
sc = StandardScaler()
x_train_scaled = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)

# 5. Fit the Logistic Regression Model
lr = LogisticRegression(max_iter=1000)
lr.fit(x_train_scaled, y_train)

print("Model trained successfully! Accuracy:", lr.score(x_test_scaled, y_test))
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

y_pred = lr.predict(x_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
import gradio as gr
import numpy as np

# 1. Define the prediction function
def predict_sleep_debt(total_sleep_hours, is_severe):
    # Prepare the input vector
    features = np.array([[total_sleep_hours, is_severe]])

    # Scale the input features using our fitted scaler
    scaled_features = sc.transform(features)

    # Make prediction
    prediction = dt.predict(scaled_features)[0]
    return prediction

# 2. Build the Gradio interface using the key inputs
interface = gr.Interface(
    fn=predict_sleep_debt,
    inputs=[
        gr.Slider(minimum=0.0, maximum=24.0, value=7.0, label="Total Sleep Hours"),
        gr.Checkbox(label="Is Severe Sleep Debt Indicator (True/False)")
    ],
    outputs=gr.Textbox(label="Predicted Sleep Debt Category"),
    title="Sleep Debt Predictor",
    description="Enter total sleep hours and check the severity indicator to predict your sleep debt status."
)

# 3. Launch the Gradio app
interface.launch(share=True)
```
